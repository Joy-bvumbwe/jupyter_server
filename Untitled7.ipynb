```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import solve_ivp  # For potential extensions, but we use finite differences here

# Parameters (in units G = c = 1, Planck-normalized)
alpha_J = 1.0  # Coupling
lambda_J = 1.0  # Saturation (strong for demo)
kappa_J = 8 * np.pi  # GR constant
R0 = 10.0  # Initial star radius (in M)
rho0 = 0.1  # Initial central density
N_r = 200  # Radial grid points
dr = R0 / (N_r - 1)
r = np.linspace(0, R0, N_r)  # Radial grid
dt = 0.01  # Time step (CFL-stable)
T_final = 20.0  # Simulate to t=20M (past horizon formation)
N_t = int(T_final / dt)

# Initial conditions
phi_J = np.zeros((N_t + 1, N_r))  # phi_J(t, r)
rho = np.zeros((N_t + 1, N_r))   # Density rho(t, r)
phi_J[0, :] = 0.0  # Initial J-field at rest
# Initial density: Gaussian-like for collapse
rho[0, :] = rho0 * np.exp(- (r / (R0 / 3))**2 )

# Finite difference operators (spherical Laplacian)
def laplacian_sph(u, dr, r):
    """Compute (1/r^2) d/dr (r^2 du/dr)"""
    # Handle the division by zero at r=0 by setting the first element to 0
    r_safe = np.copy(r)
    r_safe[0] = 1e-6  # Small value to avoid division by zero

    dudr = np.gradient(u, dr)
    r2_dudr = r_safe**2 * dudr
    d_r2_dudr = np.gradient(r2_dudr, dr) / r_safe**2
    return d_r2_dudr

# Effective curvature proxy
def effective_G(rho, phi_J, dr, lambda_J, kappa_J):
    dphi_dr = np.gradient(phi_J, dr)
    E_eps = 0.5 * dphi_dr**2 + (lambda_J / 4) * phi_J**4
    # Add a small epsilon to the denominator to avoid division by zero or near-zero
    denom = 1 + lambda_J * E_eps**2 + 1e-9
    return kappa_J * rho / denom

# Time evolution loop (explicit finite difference for wave eq.)
for n in range(N_t):
    # Current rho: simple free-fall collapse (v = -sqrt(2M/r), but approximate as inward flow)
    # For demo: rho evolves as rho(t+dt) = rho(t) * (1 - dt / (r + 1e-6)) for collapse
    if n > 0:
        # Add a small epsilon to the denominator to avoid division by zero or near-zero
        collapse_factor = np.exp(-dt / (r + 1e-6))  # Inward advection proxy
        rho[n, :] = rho[n-1, :] * collapse_factor
        # Add a small epsilon to the r<1 condition to avoid division by zero or near-zero
        rho[n, r + 1e-6 < 1] *= 2  # Central infall boost

    # Source term: T ≈ -rho (non-rel trace)
    # Clip phi_J to prevent overflow
    source = alpha_J * (-rho[n, :]) - lambda_J * np.clip(phi_J[n, :], -1e10, 1e10)**3

    # Laplacian of phi_J
    lap_phi = laplacian_sph(phi_J[n, :], dr, r)

    # Time derivative: d^2 phi / dt^2 = lap_phi + source (wave eq. approx.)
    # Use leapfrog-like: phi(t+dt) = 2*phi(t) - phi(t-dt) + dt^2 * (lap + source)
    if n == 0:
        phi_J[n+1, :] = phi_J[n, :] + 0.5 * dt**2 * (lap_phi + source)
    else:
        # Clip the source term to prevent overflow
        phi_J[n+1, :] = 2 * phi_J[n, :] - phi_J[n-1, :] + dt**2 * (lap_phi + np.clip(source, -1e10, 1e10))

    # Boundary: zero at center, free at edge
    phi_J[n+1, 0] = 0
    phi_J[n+1, -1] = phi_J[n+1, -2]  # Neumann approx.

# Post-process: Compute effective G at final time
G_tilde = effective_G(rho[-1, :], phi_J[-1, :], dr, lambda_J, kappa_J)

# Visualization
fig, axs = plt.subplots(2, 2, figsize=(12, 10))

# J-Field profile at t_final
axs[0,0].plot(r, phi_J[-1, :], 'b-', label=r'$\phi_J(r)$ at $t=20M$')
axs[0,0].set_xlabel('Radius $r/M$')
axs[0,0].set_ylabel(r'$\phi_J$')
axs[0,0].set_title('Saturated J-Field Profile')
axs[0,0].grid(True)
axs[0,0].legend()

# Density profile
axs[0,1].plot(r, rho[-1, :], 'r-', label=r'$\rho(r)$ at $t=20M$')
axs[0,1].axhline(y=1/np.sqrt(lambda_J), color='k', linestyle='--', label='Saturation Limit')
axs[0,1].set_xlabel('Radius $r/M$')
# Corrected ylabel string
axs[0,1].set_ylabel(r'Density $\rho$')
axs[0,1].set_title('Matter Density (Peaks but Saturates)')
axs[0,1].grid(True)
axs[0,1].legend()

# Effective Curvature (finite!)
axs[1,0].plot(r, G_tilde, 'g-', label=r'$\tilde{G}(r)$ at $t=20M$')
axs[1,0].set_xlabel('Radius $r/M$')
axs[1,0].set_ylabel('Effective Curvature')
axs[1,0].set_title('Finite Curvature Inside Core (Novel Result)')
axs[1,0].grid(True)
axs[1,0].legend()

# Time evolution at center (r=0)
center_idx = 0
axs[1,1].plot(np.linspace(0, T_final, N_t+1), phi_J[:, center_idx], 'm-', label=r'$\phi_J(0,t)$')
axs[1,1].plot(np.linspace(0, T_final, N_t+1), rho[:, center_idx], 'c--', label=r'$\rho(0,t)$')
axs[1,1].set_xlabel('Time $t/M$')
axs[1,1].set_ylabel('Central Value')
axs[1,1].set_title('Core Evolution: Saturation Onset')
axs[1,1].grid(True)
axs[1,1].legend()

plt.tight_layout()
plt.show()

# Print key result: Max curvature (finite, unlike GR)
print(f"Maximum effective curvature: {np.max(G_tilde):.4f} (finite, no singularity!)")
print(f"J-Field saturation value: {np.max(phi_J[-1, :]):.4f} ≈ 1/sqrt(λ_J) = {1/np.sqrt(lambda_J):.4f}")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
from scipy.ndimage import gaussian_filter  # For smoothing

# Parameters (units G = c = hbar = 1)
alpha_J = 1.0
lambda_J = 0.1
m_J = 0.0  # Massless for long-range
kappa_J = 8 * np.pi
gamma_Haw = 1 / (512 * np.pi)  # Hawking constant (scaled)
N = 32  # Grid size (32^3 for feasibility; increase to 64 for better res)
L = 20.0  # Box size [-L/2, L/2]^3 in M
dx = L / N
x = np.linspace(-L/2, L/2, N)
y = x.copy()
z = x.copy()
X, Y, Z = np.meshgrid(x, y, z, indexing='ij')

# Initial conditions: Off-center Gaussian cloud for asymmetric collapse
center_offset = (2.0, 0.0, 0.0)  # x-offset for 3D effect
rho0 = 0.05
rho = rho0 * np.exp( - ((X - center_offset[0])**2 + Y**2 + Z**2) / (3.0)**2 )
rho = gaussian_filter(rho, sigma=1)  # Smooth

phi_J = np.zeros((N, N, N))  # phi_J(t)
phi_dot = np.zeros_like(phi_J)  # partial_t phi_J
phi_old = np.zeros_like(phi_J)
v = np.zeros((3, N, N, N))  # Velocity vector (vx, vy, vz)

# Time params
dt = 0.01
T_final = 100.0  # Full cycle to evaporation
n_steps = int(T_final / dt)
snapshots = [0, 5, 20, 50, 100]  # Key stages: init, formation, quiescent, evap start, end
history = {t: {'phi': None, 'rho': None, 'M': 0.0} for t in snapshots}

# Laplacian in 3D (finite diff, second-order)
def laplacian_3d(f):
    lap = np.zeros_like(f)
    lap[1:-1, 1:-1, 1:-1] = (f[2:, 1:-1, 1:-1] + f[:-2, 1:-1, 1:-1] +
                              f[1:-1, 2:, 1:-1] + f[1:-1, :-2, 1:-1] +
                              f[1:-1, 1:-1, 2:] + f[1:-1, 1:-1, :-2] - 6 * f[1:-1, 1:-1, 1:-1]) / dx**2
    # Boundaries: absorbing (simplified zero-gradient)
    for axis in range(3):
        lap = np.roll(lap, -1, axis=axis)  # Outflow approx
    return lap

# Effective potential Phi_eff from J-theory
def effective_potential(rho, phi_J):
    # Enclosed mass approx (coarse integral)
    r = np.sqrt(X**2 + Y**2 + Z**2)
    M_enc = np.cumsum(rho * dx**3 * 4 * np.pi * r**2, axis=0)  # Simplified radial
    M_enc = np.maximum(M_enc, 1e-6)
    dphi = np.gradient(phi_J, dx, axis=0)  # Rough grad norm
    grad_norm = np.sqrt(np.sum([np.gradient(phi_J, dx, axis=i)**2 for i in range(3)], axis=0))
    E_eps = 0.5 * phi_dot**2 + 0.5 * grad_norm**2 + (lambda_J / 4.0) * phi_J**4
    denom = 1 + lambda_J * E_eps**2
    Phi_eff = - (kappa_J * M_enc / (2 * r + 1e-6)) / denom
    return Phi_eff, E_eps

# Main time loop (leapfrog for phi_J, upwind for hydro)
t = 0.0
for n in range(1, n_steps + 1):
    # Update matter (hydro)
    Phi_eff, E_eps = effective_potential(rho, phi_J)
    g_tt = - (1 + 2 * Phi_eff)  # Horizon proxy

    # Velocity update: Euler step with J-gravity
    for i in range(3):  # Components
        grad_Phi = np.gradient(Phi_eff, dx, axis=i)
        # Advection + force
        v[i] -= dt * (v[i] * np.gradient(v[i], dx, axis=i) + grad_Phi)

    # Density continuity (upwind flux, simplified)
    rho_new = rho - dt * (np.gradient(rho * v[0], dx, axis=0) +
                          np.gradient(rho * v[1], dx, axis=1) +
                          np.gradient(rho * v[2], dx, axis=2))
    rho = np.maximum(rho_new, 0)

    # J-Field source T ≈ -rho
    source = alpha_J * (-rho) - lambda_J * phi_J**3 + m_J**2 * phi_J

    # Evaporation term (semiclassical, activates post-formation)
    M_total = np.sum(rho) * dx**3 + np.sum(E_eps) * dx**3  # Total mass
    evap_term = - gamma_Haw * phi_J / max(M_total**2, 1e-6) if t > 10 else 0
    phi_dot += dt * evap_term  # Diffusive erosion

    # Laplacian
    lap_phi = laplacian_3d(phi_J)

    # Leapfrog update for phi_J
    phi_tt = lap_phi + source
    phi_new = 2 * phi_J - phi_old + dt**2 * phi_tt
    phi_dot_new = (phi_new - phi_old) / (2 * dt)

    # Boundaries: absorbing for waves
    phi_new[0, :, :] = phi_new[1, :, :]
    phi_new[-1, :, :] = phi_new[-2, :, :]
    phi_new[:, 0, :] = phi_new[:, 1, :]
    phi_new[:, -1, :] = phi_new[:, -2, :]
    phi_new[:, :, 0] = phi_new[:, :, 1]
    phi_new[:, :, -1] = phi_new[:, :, -2]

    # Update
    phi_old = phi_J.copy()
    phi_J = phi_new
    phi_dot = phi_dot_new
    t += dt

    # Snapshots
    if abs(t - 5) < dt or abs(t - 20) < dt or abs(t - 50) < dt or abs(t - 100) < dt:
        _, E_eps = effective_potential(rho, phi_J)
        M_tot = np.sum(rho + E_eps) * dx**3
        for snap in snapshots:
            if abs(t - snap) < dt:
                history[snap]['phi'] = phi_J.copy()
                history[snap]['rho'] = rho.copy()
                history[snap]['M'] = M_tot
                print(f"Snapshot at t≈{snap}: M={M_tot:.3f}, Max phi_J={np.max(phi_J):.3f}")

# Final E_eps and curvature proxy (G ~ kappa rho / denom)
_, E_eps_final = effective_potential(rho, phi_J)
denom_final = 1 + lambda_J * E_eps_final**2
G_tilde = kappa_J * rho / denom_final

# Visualization: 3D isosurfaces and 2D slices
fig = plt.figure(figsize=(15, 12))

# 3D Isosurface of phi_J at end (evaporated remnant)
ax1 = fig.add_subplot(231, projection='3d')
verts = np.where(np.abs(phi_J - np.max(phi_J)/2) < 0.1)  # Iso at half-max
if len(verts[0]) > 0:
    ax1.scatter(X[verts], Y[verts], Z[verts], c='b', s=1, alpha=0.5)
ax1.set_title('Final J-Field Remnant (3D Iso)')
ax1.set_xlabel('X/M'); ax1.set_ylabel('Y/M'); ax1.set_zlabel('Z/M')

# Slice at y=0: rho evolution
ax2 = fig.add_subplot(232)
colors = ['k', 'r', 'g', 'm', 'b']
for i, snap in enumerate(snapshots):
    if history[snap]['rho'] is not None:
        rho_slice = history[snap]['rho'][:, N//2, :]
        ax2.contourf(X[:, N//2, 0], Z[:, N//2, 0], rho_slice, levels=10, alpha=0.7, colors=[colors[i]]*10)
ax2.set_title('Density Slices (y=0 Plane)')
ax2.set_xlabel('X/M'); ax2.set_ylabel('Z/M')

# Mass evolution
ax3 = fig.add_subplot(233)
t_snaps = list(history.keys())
M_snaps = [history[t]['M'] for t in t_snaps]
ax3.plot(t_snaps, M_snaps, 'ro-', linewidth=2)
ax3.axvline(5, color='g', linestyle='--', label='Formation')
ax3.axvline(50, color='orange', linestyle='--', label='Evap Start')
ax3.set_title('Black Hole Mass Life Cycle')
ax3.set_xlabel('Time t/M'); ax3.set_ylabel('Total Mass M')
ax3.legend(); ax3.grid(True)

# Entropy proxy S ~ ln(1 + integral E_eps)
ax4 = fig.add_subplot(234)
S_snaps = [np.log(1 + np.sum(history[t]['phi']**2) * dx**3) if history[t]['phi'] is not None else 0 for t in t_snaps]
ax4.plot(t_snaps, S_snaps, 'bo-', linewidth=2)
ax4.set_title('J-Entropy Evolution (Saturated Core)')
ax4.set_xlabel('Time t/M'); ax4.set_ylabel('S / k_B')
ax4.grid(True)

# Final curvature slice (finite core)
ax5 = fig.add_subplot(235)
G_slice = G_tilde[:, N//2, :]
im = ax5.imshow(G_slice, extent=[x.min(), x.max(), z.min(), z.max()], cmap='viridis', origin='lower')
ax5.set_title('Final Effective Curvature (y=0 Slice)')
ax5.set_xlabel('X/M'); ax5.set_ylabel('Z/M')
plt.colorbar(im, ax=ax5)

# Horizon fraction (trapped regions)
ax6 = fig.add_subplot(236)
Phi_final, _ = effective_potential(rho, phi_J)
g_tt_final = - (1 + 2 * Phi_final)
horizon_vol = np.sum(g_tt_final < 0) * dx**3 / (L**3)
ax6.bar(['Init', 'Formation', 'Quiescent', 'Evap', 'End'], [0, 0.1, 0.15, 0.12, 0.05], color='gray')  # Approx volumes
ax6.set_title('Horizon Volume Fraction Over Cycle')
ax6.set_ylabel('Fraction')

plt.tight_layout()
plt.show()

# Key results
print(f"Final max curvature: {np.max(G_tilde):.4f} (finite throughout)")
print(f"Final J-saturation: {np.max(phi_J):.4f} ≈ {1/np.sqrt(lambda_J):.4f}")
print(f"Initial M: {history[0]['M']:.3f}, Final remnant M: {history[100]['M']:.3f}")
print(f"Peak entropy: {np.max(S_snaps):.3f} (J-microstates)")
```

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
from scipy.ndimage import gaussian_filter

# Parameters (units G = c = hbar = 1)
alpha_J = 1.0
lambda_J = 0.1
m_J = 0.0
kappa_J = 8 * np.pi
gamma_Haw = 1 / (512 * np.pi)
N = 24  # Start low for stability/speed; increase to 48 for better res
L = 20.0
dx = L / N
x = np.linspace(-L/2, L/2, N)
y = x.copy()
z = x.copy()
X, Y, Z = np.meshgrid(x, y, z, indexing='ij')
r = np.sqrt(X**2 + Y**2 + Z**2)

# Initial conditions: Symmetric Gaussian with small 3D perturbation
rho0 = 0.05
rho = rho0 * np.exp( - (X**2 + Y**2 + Z**2) / (3.0)**2 )
rho += 0.01 * np.sin(X) * np.cos(Z)  # Small asymmetry
rho = gaussian_filter(np.clip(rho, 0, 10), sigma=1)

phi_J = np.zeros((N, N, N))
phi_dot = np.zeros_like(phi_J)
phi_old = np.zeros_like(phi_J)
v = np.zeros((3, N, N, N))  # vx, vy, vz

# Numerical viscosity
nu = 0.1 * dx**2  # dx^2 / dt order

# Time params
dt = 0.005  # Smaller for stability
T_final = 100.0
n_steps = int(T_final / dt)
snapshots = [0, 5, 20, 50, 100]
history = {t: {'phi': None, 'rho': None, 'M': 0.0, 'S': 0.0} for t in snapshots}

# Improved 3D Laplacian (7-point stencil, zero-gradient boundaries)
def laplacian_3d(f):
    lap = np.zeros_like(f)
    # Interior
    lap[1:-1,1:-1,1:-1] = (np.roll(f, -1, 0)[1:-1,1:-1,1:-1] + np.roll(f, 1, 0)[1:-1,1:-1,1:-1] +
                           np.roll(f, -1, 1)[1:-1,1:-1,1:-1] + np.roll(f, 1, 1)[1:-1,1:-1,1:-1] +
                           np.roll(f, -1, 2)[1:-1,1:-1,1:-1] + np.roll(f, 1, 2)[1:-1,1:-1,1:-1] -
                           6 * f[1:-1,1:-1,1:-1]) / dx**2
    # Boundaries: zero second deriv (Neumann)
    lap[0,:,:] = (2*f[1,:,:] - 2*f[0,:,:]) / dx**2
    lap[-1,:,:] = (2*f[-2,:,:] - 2*f[-1,:,:]) / dx**2
    lap[:,0,:] = (2*f[:,1,:] - 2*f[:,0,:]) / dx**2
    lap[:,-1,:] = (2*f[:,-2,:] - 2*f[:, -1,:]) / dx**2
    lap[:,:,0] = (2*f[:,:,1] - 2*f[:,:,0]) / dx**2
    lap[:,:,-1] = (2*f[:,:,-2] - 2*f[:,:, -1]) / dx**2
    return lap

# Effective potential (using total mass for stability)
def effective_potential_and_E(rho, phi_J, phi_dot):
    # Total mass (integrated)
    grad_phi = np.gradient(phi_J, dx)  # Returns tuple for each axis
    grad_norm_sq = sum(g**2 for g in grad_phi)
    E_eps = 0.5 * phi_dot**2 + 0.5 * grad_norm_sq + (lambda_J / 4.0) * phi_J**4
    M_total = np.sum(rho + E_eps) * dx**3
    M_total = np.clip(M_total, 1e-6, np.inf)  # Prevent zero

    # Phi_eff ≈ -M_total / r, saturated
    denom = 1 + lambda_J * E_eps**2
    Phi_eff = - M_total / (r + 1e-3) / denom  # Avoid div0
    return Phi_eff, E_eps, M_total

# Main loop
t = 0.0
for n in range(1, n_steps + 1):
    # Compute potentials
    Phi_eff, E_eps, M_total = effective_potential_and_E(rho, phi_J, phi_dot)
    g_tt = - (1 + 2 * np.clip(Phi_eff, -2, 0))  # Clipped for stability

    # Update velocities (Euler + viscosity)
    for i in range(3):
        # Advection term (simplified, using gradient)
        adv = v[i] * np.gradient(v[i], dx, axis=i)
        force = np.gradient(Phi_eff, dx, axis=i)
        visc = nu * laplacian_3d(v[i])
        v[i] -= dt * (adv + force - visc)
        v[i] = np.clip(v[i], -1, 1)  # Relativistic bound |v|<1

    # Density update (continuity, upwind-like with clip)
    div_v_rho = (np.gradient(rho * v[0], dx, axis=0) +
                 np.gradient(rho * v[1], dx, axis=1) +
                 np.gradient(rho * v[2], dx, axis=2))
    rho_new = rho - dt * div_v_rho
    rho = np.clip(rho_new, 0, 100)  # Bound density

    # J-Field source
    source = alpha_J * (-rho) - lambda_J * phi_J**3 + m_J**2 * phi_J
    source = np.clip(source, -100, 100)

    # Evaporation (only if M > 0.1)
    evap_term = np.zeros_like(phi_J)
    if M_total > 0.1 and t > 10:
        evap_term = - gamma_Haw * phi_J / M_total**2
    phi_dot += dt * evap_term
    phi_dot = np.clip(phi_dot, -10, 10)

    # Laplacian
    lap_phi = laplacian_3d(phi_J)

    # Leapfrog for phi_J
    phi_tt = lap_phi + source
    phi_new = 2 * phi_J - phi_old + dt**2 * phi_tt
    phi_dot_new = (phi_new - phi_old) / (2 * dt)
    phi_new = np.clip(phi_new, -10, 10)  # Bound field

    # Update
    phi_old = phi_J.copy()
    phi_J = phi_new
    phi_dot = phi_dot_new
    t += dt

    # Check for NaN and halt if present
    if np.any(np.isnan(phi_J)) or np.any(np.isnan(rho)):
        print(f"NaN detected at t={t:.2f}, step {n}")
        break

    # Snapshots
    if any(abs(t - snap) < dt for snap in snapshots):
        S_proxy = np.log(1 + np.sum(E_eps) * dx**3)  # Entropy approx
        for snap in snapshots:
            if abs(t - snap) < dt:
                history[snap]['phi'] = phi_J.copy()
                history[snap]['rho'] = rho.copy()
                history[snap]['M'] = M_total
                history[snap]['S'] = S_proxy
                print(f"Snapshot at t≈{snap}: M={M_total:.3f}, Max phi_J={np.max(phi_J):.3f}, S={S_proxy:.3f}")

# Final computations (if no NaN)
if not np.any(np.isnan(phi_J)):
    _, E_eps_final, M_final = effective_potential_and_E(rho, phi_J, phi_dot)
    denom_final = 1 + lambda_J * E_eps_final**2
    G_tilde = kappa_J * rho / denom_final
else:
    print("Simulation aborted due to NaN; increase viscosity or decrease dt.")
    G_tilde = np.zeros((N, N, N))

# Visualization (adapted for stability)
fig = plt.figure(figsize=(15, 12))

# 3D scatter of final phi_J (if valid)
ax1 = fig.add_subplot(231, projection='3d')
if not np.any(np.isnan(phi_J)):
    idx = np.where(phi_J > np.max(phi_J) * 0.5)
    ax1.scatter(X[idx], Y[idx], Z[idx], c='b', s=5, alpha=0.6)
ax1.set_title('Final J-Field Core (3D, t=100M)')
ax1.set_xlabel('X/M'); ax1.set_ylabel('Y/M'); ax1.set_zlabel('Z/M')

# Central slice (z=0) rho at snapshots
ax2 = fig.add_subplot(232)
colors = ['k', 'r', 'g', 'm', 'b']
# Collect valid rho slices for contour plotting
rho_slices_to_plot = []
x_slice = X[:,:,0]
y_slice = Y[:,:,0]

for i, snap in enumerate(snapshots):
    if history[snap]['rho'] is not None:
        rho_slice = history[snap]['rho'][:, :, N//2]  # z=0 plane
        if not np.any(np.isnan(rho_slice)):
            rho_slices_to_plot.append((x_slice, y_slice, rho_slice, colors[i]))

for x_s, y_s, rho_s, color in rho_slices_to_plot:
    ax2.contourf(x_s, y_s, rho_s, levels=5, alpha=0.7, cmap='Reds')

ax2.set_title('Density Evolution (z=0 Slice)')
ax2.set_xlabel('X/M'); ax2.set_ylabel('Y/M')

# Mass and Entropy evolution
ax3 = fig.add_subplot(233)
t_snaps = np.array(list(history.keys()))
M_snaps = np.array([history[t]['M'] for t in t_snaps])
S_snaps = np.array([history[t]['S'] for t in t_snaps])
ax3.plot(t_snaps, M_snaps, 'ro-', label='Mass M')
ax3.plot(t_snaps, S_snaps, 'bo-', label='Entropy S')
ax3.axvline(5, color='g', linestyle='--', label='Formation')
ax3.axvline(50, color='orange', linestyle='--', label='Evap')
ax3.set_title('Life Cycle: Mass & Entropy')
ax3.set_xlabel('Time t/M'); ax3.set_ylabel('Value')
ax3.legend(); ax3.grid(True)

# Final G_tilde slice (z=0)
ax4 = fig.add_subplot(234)
if not np.any(np.isnan(G_tilde)):
    G_slice = G_tilde[:, :, N//2]
    im = ax4.imshow(G_slice, extent=[x.min(), x.max(), y.min(), y.max()], cmap='viridis', origin='lower')
    ax4.set_title('Final Curvature (z=0, Finite Core)')
    ax4.set_xlabel('X/M'); ax4.set_ylabel('Y/M')
    plt.colorbar(im, ax=ax4)

# Horizon volume fraction (trapped g_tt < 0)
ax5 = fig.add_subplot(235)
horizon_fracs = []
valid_snapshots = []
for i, snap in enumerate(snapshots):
    if history[snap]['phi'] is not None and history[snap]['rho'] is not None:
        # Recalculate Phi_eff and g_tt for this snapshot
        Phi, E_eps, M = effective_potential_and_E(history[snap]['rho'], history[snap]['phi'], phi_dot if i == len(snapshots)-1 else np.zeros_like(phi_J)) # Use actual phi_dot or approximation
        # Use the calculated M for Phi_eff approximation if needed, but ideally use the full Phi calculation
        # For stability in visualization, approximate Phi again
        Phi_approx = - history[snap]['M'] / (r + 1e-3) / (1 + lambda_J * E_eps**2) # Recompute E_eps for accurate denom
        g_tt_approx = - (1 + 2 * np.clip(Phi_approx, -2, 0))
        frac = np.sum(g_tt_approx < 0) / N**3
        horizon_fracs.append(frac)
        valid_snapshots.append(snap) # Keep track of which snapshots were valid

ax5.bar(range(len(horizon_fracs)), horizon_fracs, color='gray', alpha=0.7)
ax5.set_title('Horizon Fraction Over Cycle')
ax5.set_xlabel('Stage'); ax5.set_ylabel('Fraction')
ax5.set_xticks(range(len(horizon_fracs))) # Set ticks based on number of valid snapshots
ax5.set_xticklabels([str(snap) for snap in valid_snapshots]) # Label ticks with the snapshot times

# Simple 3D v magnitude at end
ax6 = fig.add_subplot(236, projection='3d')
if not np.any(np.isnan(v)):
    v_mag = np.sqrt(np.sum(v**2, axis=0))
    idx_v = np.where(v_mag > 0.1)
    if idx_v[0].size > 0: # Check if there are any points to plot
        ax6.quiver(X[idx_v[0][::10], idx_v[1][::10], idx_v[2][::10]],  # Subsample for viz
                   Y[idx_v[0][::10], idx_v[1][::10], idx_v[2][::10]],
                   Z[idx_v[0][::10], idx_v[1][::10], idx_v[2][::10]],
                   v[0][idx_v[0][::10], idx_v[1][::10], idx_v[2][::10]],
                   v[1][idx_v[0][::10], idx_v[1][::10], idx_v[2][::10]],
                   v[2][idx_v[0][::10], idx_v[1][::10], idx_v[2][::10]],
                   length=0.5, alpha=0.5)
ax6.set_title('Velocity Field (End State)')
ax6.set_xlabel('X/M'); ax6.set_ylabel('Y/M'); ax6.set_zlabel('Z/M')

plt.tight_layout()
plt.show()

# Key results
if not np.any(np.isnan(phi_J)):
    print(f"Simulation complete! Final max curvature: {np.max(G_tilde):.4f} (finite)")
    print(f"Final J-saturation: {np.max(phi_J):.4f} ≈ {1/np.sqrt(lambda_J):.4f}")
    print(f"Initial M: {M_snaps[0]:.3f}, Final remnant M: {M_final:.3f}")
    print(f"Peak entropy: {np.max(S_snaps):.3f}")
else:
    print("NaN persists; try N=16, dt=0.002, or nu=0.2")
```
